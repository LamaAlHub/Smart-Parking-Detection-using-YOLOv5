```python
import torch 
import cv2
```

```python
import pathlib
temp = pathlib.PosixPath 
pathlib.PosixPath = pathlib.WindowsPath
```

```python
model = torch.hub.load("ultralytics/yolov5","custom",path = r"C:\Users\imyla\Downloads\CarSpaceYOLO\yolov5\runs\train\exp\weights\best.pt",force_reload = True)
```

```python
import numpy as np
```

```python
cap = cv2.VideoCapture(r"C:\Users\imyla\Downloads\istockphoto-2204962548-640_adpp_is.mp4")
while True:
    ret,frame = cap.read()
    frame = cv2.resize(frame,(600,300))
    if not cap.isOpened():
        print("Error: Could not open video stream.")
        exit()
   
    if not ret:
        print("Error: Can't receive frame. Exiting...")
        break 
    model.conf = 0.6
    result = model(frame)
    
    emptySpaceCount = 0
    occupiedSpaceCount =0

    for detection in result.xyxy[0]:
        class_id = int(detection[5])
        class_name = model.names[class_id]
        if class_name == "empty":
            emptySpaceCount += 1
        elif class_name == "occupied":
            occupiedSpaceCount += 1
    
    annotated_img = np.squeeze(result.render())
    annotated_img = cv2.rectangle(annotated_img,(0,10),(150,80),(0,0,0),-1)
    annotated_img = cv2.putText(annotated_img,"parking analyis",(5,30),cv2.FONT_HERSHEY_SIMPLEX,0.5,(255,255,255),2)
    annotated_img = cv2.putText(annotated_img,"empty parking: "+ str(emptySpaceCount),(5,50),cv2.FONT_HERSHEY_SIMPLEX,0.5,(255,255,255),2)
    annotated_img = cv2.putText(annotated_img,"occupied parking: "+ str(occupiedSpaceCount),(5,70),cv2.FONT_HERSHEY_SIMPLEX,0.5,(255,255,255),2)


    cv2.imshow('Car space YOLO', annotated_img)
    

    # 4. Stop the loop if the 'q' key is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
```
